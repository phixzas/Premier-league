```python
import schedule
import time
import subprocess
from datetime import datetime, timedelta
import requests
import os

def retrain():
    print(f"[{datetime.now()}] Monday 10am - Retraining V9...")
    os.makedirs("data", exist_ok=True)
    try:
        url = "https://www.football-data.co.uk/mmz4281/2526/E0.csv"
        r = requests.get(url, timeout=20)
        open("data/2526.csv","wb").write(r.content)
        print("Downloaded latest 2627.csv")
    except Exception as e:
        print(f"Download failed: {e}, using old file")
    subprocess.run(["python", "train_v9.py"])
    print("Retrain done!")

schedule.every().monday.at("10:00").do(retrain)
print("Auto-Retrain V9 running... Waiting for Monday 10:00. Keep this running.")
# retrain() # Remove # to test now

while True:
    schedule.run_pending()
    time.sleep(60)
```
